```python
engine = create_engine(f'postgresql://postgres:samadhi@localhost:5432/postgres')

try:
    with engine.connect() as conn:
        result = conn.execute(text("SELECT 1"))
        print("Connected to PostgreSQL successfully!")
except Exception as e:
    print(f"Connection failed: {e}")
```

```python
#create database
with engine.connect() as conn:
    conn.execute(text("COMMIT"))

    # Check if database exists
    result = conn.execute(text("SELECT 1 FROM pg_database WHERE datname = 'ecommerce_db'"))
    exists = result.fetchone()
    
    if not exists:
        conn.execute(text("CREATE DATABASE ecommerce_db"))
        print("Database 'ecommerce_db' created!")
    else:
        print("Database 'ecommerce_db' already exists")
```

```python
#connect to database
engine = create_engine('postgresql://postgres:samadhi@localhost:5432/ecommerce_db')

# Test connection
try:
    with engine.connect() as conn:
        result = conn.execute(text("SELECT 1"))
        print("Connected to ecommerce_db successfully!")
except Exception as e:
    print(f"Connection failed: {e}")
```

```python
#load tables

files_to_load = [
    'online_retail_cleaned.csv',
    'customer_segments.csv', 
    'clv_predictions.csv',
    'features_daily_sales.csv'
]

for file in files_to_load:
    print(f"\nLoading {file}...")
    
    df = pd.read_csv(f'{file}')
    print(f"   Rows: {len(df)}")
    
    table_name = file.replace('.csv', '').replace('online_retail_cleaned', 'transactions')
    
    df.to_sql(table_name, engine, if_exists='replace', index=False)
    print(f" Loaded to table: {table_name}")

print("\nAll data loaded successfully!")
```
